```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

digits = load_digits()

X = digits.data
y = digits.target

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

modelo = GaussianNB()

modelo.fit(X_treino, y_treino)

predicoes = modelo.predict(X_teste)

acuracia = accuracy_score(y_teste, predicoes)

print("Acurácia:", round(acuracia * 100, 2), "%")

matriz = confusion_matrix(y_teste, predicoes)

plt.figure(figsize=(8,6))
sns.heatmap(matriz, annot=True, fmt="d")
plt.title("Matriz de Confusão")
plt.xlabel("Predito")
plt.ylabel("Real")
plt.show()

indice = 10

amostra = X_teste[indice].reshape(1, -1)

resultado = modelo.predict(amostra)

print("\nPredição realizada:")
print("Número previsto:", resultado[0])

print("Número real:", y_teste[indice])
```
